```python
import pickle
import numpy as np
import pandas as pd

# 1. Load your trained production pipeline back into memory
with open('airbnb_model_pipeline.pkl', 'rb') as file:
    model_pipeline = pickle.load(file)

print("✅ Saved pipeline loaded successfully!")

# 2. Simulate a completely new listing entered by a user
# Notice we provide raw text categories ('Manhattan', 'Private room')
new_listing_data = pd.DataFrame([{
    'latitude': 40.7589, 
    'longitude': -73.9851, 
    'minimum_nights': 3,
    'number_of_reviews': 15, 
    'reviews_per_month': 1.2,
    'calculated_host_listings_count': 1, 
    'availability_365': 180,
    'neighbourhood_group': 'Manhattan',  # Categorical string
    'room_type': 'Private room'         # Categorical string
}])

print("\n--- Testing Live Web Input ---")
print(new_listing_data)

# 3. Pass the raw input directly through the pipeline
# The pipeline automatically handles the One-Hot Encoding behind the scenes
predicted_log_price = model_pipeline.predict(new_listing_data)

# 4. Reverse the log transformation to get the final price in real US Dollars
final_usd_price = np.expm1(predicted_log_price)[0]

print(f"\n🎉 Predicted Price Output: ${final_usd_price:.2f} per night")
```
