### Task 2 - Vectorized Arithmetic and Linear Algebra

```python
# Import NumPy library
import numpy as np

# Create two matrices
A = np.array([[2, 3, 4],
              [1, 5, 6],
              [7, 8, 9]])

B = np.array([[1, 2, 3],
              [4, 5, 6],
              [7, 8, 10]])

# Display matrices
print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)


# 1. Matrix Addition
print("\nMatrix Addition (A + B):")
print(A + B)


# 2. Element-wise Multiplication
print("\nElement-wise Multiplication (A * B):")
print(A * B)


# 3. Matrix Multiplication using @ operator
print("\nMatrix Multiplication (A @ B):")
print(A @ B)


# 4. Transpose of Matrix A
print("\nTranspose of Matrix A:")
print(A.T)


# 5. Determinant of Matrix A
det_A = np.linalg.det(A)

print("\nDeterminant of Matrix A:")
print(det_A)


# 6. Finding inverse if matrix is invertible

if det_A != 0:
    inverse_A = np.linalg.inv(A)

    print("\nInverse of Matrix A:")
    print(inverse_A)


    # Verify A @ A^-1 = Identity Matrix
    result = A @ inverse_A

    print("\nA @ A^-1:")
    print(result)


    print("\nVerification using np.allclose():")
    print(np.allclose(result, np.eye(3)))

else:
    print("\nMatrix A is not invertible")
```

### Task 3 - Normal Distribution and Histogram


```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Choose mean and standard deviation
mean = 50
std = 10

# Generate 1000 random values from normal distribution
data = np.random.normal(mean, std, 1000)

# Calculate sample mean and sample standard deviation
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)   # ddof=1 gives sample standard deviation

# Display results
print("Chosen Mean:", mean)
print("Chosen Standard Deviation:", std)

print("\nSample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)


# Compare chosen and sample values
print("\nDifference between chosen mean and sample mean:",
      abs(mean - sample_mean))

print("Difference between chosen std and sample std:",
      abs(std - sample_std))


# Plot histogram
plt.figure(figsize=(8,5))

plt.hist(data, bins=30)

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram of Normal Distribution")

plt.show()
```

### Task 5 - Filtering and Querying


```python
import pandas as pd

# Load Titanic dataset
df = pd.read_csv("train.csv")


# 1. Male passengers older than 50
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

print("1. Male passengers older than 50:")
print("Count =", len(male_over_50))


# 2. Female first-class passengers
female_first_class = df[
    (df["Sex"] == "female") & (df["Pclass"] == 1)
]

# Number of female first-class passengers
count_female_first_class = len(female_first_class)

# Percentage survived
survival_percentage = female_first_class["Survived"].mean() * 100

print("\n2. Female first-class passengers:")
print("Count =", count_female_first_class)
print("Survival percentage =", survival_percentage, "%")


# 3. Age 20-40, Fare above overall median, and survived
fare_median = df["Fare"].median()

condition_3 = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > fare_median) &
    (df["Survived"] == 1)
]

print("\n3. Age 20-40, Fare above median, and survived:")
print("Count =", len(condition_3))


# 4. Travelling alone, age < 30, and did not survive
condition_4 = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("\n4. Travelling alone, age < 30, and did not survive:")
print("Count =", len(condition_4))


# 5. Embarked = S, Pclass 2 or 3,
#    and Fare above Southampton median

southampton_median_fare = df[df["Embarked"] == "S"]["Fare"].median()

condition_5 = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]

print("\n5. Embarked = S, Pclass 2 or 3, and Fare above Southampton median:")
print("Southampton median fare =", southampton_median_fare)
print("Count =", len(condition_5))
```

### Task 6 - groupby() and Aggregation

```python
# Import Pandas
import pandas as pd

# Load Titanic dataset
df = pd.read_csv("train.csv")


# ============================================================
# 1. SURVIVAL RATE BY SEX
# ============================================================

survival_by_sex = (
    df.groupby("Sex")["Survived"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index()
)

survival_by_sex.columns = ["Sex", "Survival Rate (%)"]

print("1. Survival Rate by Sex")
display(survival_by_sex)


# ============================================================
# 2. SURVIVAL RATE BY PCLASS
# ============================================================

survival_by_pclass = (
    df.groupby("Pclass")["Survived"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index()
)

survival_by_pclass.columns = ["Pclass", "Survival Rate (%)"]

print("2. Survival Rate by Pclass")
display(survival_by_pclass)


# ============================================================
# 3. AVERAGE AGE AND FARE BY PCLASS
# ============================================================

avg_age_fare = (
    df.groupby("Pclass")[["Age", "Fare"]]
    .mean()
    .round(2)
    .reset_index()
)

avg_age_fare.columns = ["Pclass", "Average Age", "Average Fare"]

print("3. Average Age and Fare by Pclass")
display(avg_age_fare)


# ============================================================
# 4. PASSENGER COUNT AND SURVIVAL RATE BY SEX-PCLASS
# ============================================================

sex_pclass = (
    df.groupby(["Sex", "Pclass"])
    .agg(
        Passenger_Count=("PassengerId", "count"),
        Survival_Rate=("Survived", "mean")
    )
    .reset_index()
)

# Convert survival rate to percentage
sex_pclass["Survival_Rate"] = (
    sex_pclass["Survival_Rate"] * 100
).round(2)

sex_pclass.columns = [
    "Sex",
    "Pclass",
    "Passenger Count",
    "Survival Rate (%)"
]

print("4. Passenger Count and Survival Rate by Sex-Pclass")
display(sex_pclass)


# ============================================================
# 5. PASSENGER COUNT, AVERAGE FARE AND SURVIVAL RATE BY EMBARKED
# ============================================================

embarked_stats = (
    df.groupby("Embarked")
    .agg(
        Passenger_Count=("PassengerId", "count"),
        Average_Fare=("Fare", "mean"),
        Survival_Rate=("Survived", "mean")
    )
    .reset_index()
)

# Round average fare
embarked_stats["Average_Fare"] = (
    embarked_stats["Average_Fare"].round(2)
)

# Convert survival rate to percentage
embarked_stats["Survival_Rate"] = (
    embarked_stats["Survival_Rate"] * 100
).round(2)

embarked_stats.columns = [
    "Embarked",
    "Passenger Count",
    "Average Fare",
    "Survival Rate (%)"
]

print("5. Passenger Count, Average Fare and Survival Rate by Embarked")
display(embarked_stats)
```

### Task 7 - Missing Values and Fare Outliers


```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load Titanic dataset
df = pd.read_csv("train.csv")


# ============================================================
# 1. MISSING VALUE COUNT AND PERCENTAGE FOR EVERY COLUMN
# ============================================================

missing_count = df.isnull().sum()

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Column": df.columns,
    "Missing Count": missing_count.values,
    "Missing Percentage (%)": missing_percentage.round(2).values
})

print("1. MISSING VALUES")
display(missing_table)


# ============================================================
# 2. PLOT MISSING VALUES AS A BAR CHART
# ============================================================

plt.figure(figsize=(10, 5))

plt.bar(missing_table["Column"], missing_table["Missing Count"])

plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.title("Missing Values in Each Column")

plt.xticks(rotation=45)

plt.show()


# ============================================================
# 3. FILL MISSING AGE VALUES WITH MEAN AGE
# ============================================================

# Count missing Age values BEFORE filling
age_missing_before = df["Age"].isnull().sum()

# Calculate mean Age
mean_age = df["Age"].mean()

# Fill missing Age values with mean
df_mean = df.copy()
df_mean["Age"] = df_mean["Age"].fillna(mean_age)

# Count missing Age values AFTER filling
age_missing_after = df_mean["Age"].isnull().sum()

mean_imputation_table = pd.DataFrame({
    "Method": ["Mean Imputation"],
    "Mean Age Used": [round(mean_age, 2)],
    "Missing Before": [age_missing_before],
    "Missing After": [age_missing_after]
})

print("2. MEAN IMPUTATION FOR AGE")
display(mean_imputation_table)


# ============================================================
# 4. TRY DIFFERENT IMPUTATION METHODS
#    Mean, Median, Mode and Random Value
# ============================================================

# Create separate copies
df_mean = df.copy()
df_median = df.copy()
df_mode = df.copy()
df_random = df.copy()


# -------------------------
# Mean Imputation
# -------------------------

mean_age = df["Age"].mean()

df_mean["Age"] = df_mean["Age"].fillna(mean_age)


# -------------------------
# Median Imputation
# -------------------------

median_age = df["Age"].median()

df_median["Age"] = df_median["Age"].fillna(median_age)


# -------------------------
# Mode Imputation
# -------------------------

mode_age = df["Age"].mode()[0]

df_mode["Age"] = df_mode["Age"].fillna(mode_age)


# -------------------------
# Random Value Imputation
# -------------------------

# Set seed so results are reproducible
np.random.seed(42)

# Get non-missing Age values
available_ages = df["Age"].dropna()

# Find missing Age positions
missing_positions = df_random["Age"].isnull()

# Randomly select Age values from existing non-missing values
random_ages = np.random.choice(
    available_ages,
    size=missing_positions.sum()
)

# Replace missing values
df_random.loc[missing_positions, "Age"] = random_ages


# ============================================================
# 5. COMPARISON TABLE FOR IMPUTATION METHODS
# ============================================================

imputation_table = pd.DataFrame({
    "Method": [
        "Mean",
        "Median",
        "Mode",
        "Random Value"
    ],

    "Value Used": [
        round(mean_age, 2),
        round(median_age, 2),
        round(mode_age, 2),
        "Random values from Age column"
    ],

    "Missing Before": [
        df["Age"].isnull().sum()
    ] * 4,

    "Missing After": [
        df_mean["Age"].isnull().sum(),
        df_median["Age"].isnull().sum(),
        df_mode["Age"].isnull().sum(),
        df_random["Age"].isnull().sum()
    ]
})

print("3. COMPARISON OF AGE IMPUTATION METHODS")
display(imputation_table)


# ============================================================
# 6. FARE OUTLIERS USING IQR
# ============================================================

# Calculate Q1 and Q3
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

# Create outlier summary table
outlier_table = pd.DataFrame({
    "Measure": [
        "Q1",
        "Q3",
        "IQR",
        "Lower Bound",
        "Upper Bound",
        "Number of Outliers"
    ],

    "Value": [
        round(Q1, 2),
        round(Q3, 2),
        round(IQR, 2),
        round(lower_bound, 2),
        round(upper_bound, 2),
        len(fare_outliers)
    ]
})

print("4. FARE OUTLIER ANALYSIS USING IQR")
display(outlier_table)
```

### Task 8 - Features and Pivot Table


```python
# Import Pandas
import pandas as pd

# Load Titanic dataset
df = pd.read_csv("train.csv")


# ============================================================
# 1. CREATE FamilySize
# ============================================================

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1


# ============================================================
# 2. CREATE IsAlone
# ============================================================

# If FamilySize = 1, passenger is travelling alone
# Otherwise, passenger is not travelling alone

df["IsAlone"] = (df["FamilySize"] == 1).astype(int)


# ============================================================
# 3. DISPLAY THE NEW FEATURES
# ============================================================

print("1. NEW FEATURES: FamilySize AND IsAlone")

family_table = df[
    ["PassengerId", "SibSp", "Parch", "FamilySize", "IsAlone"]
].head(10)

display(family_table)


# ============================================================
# 4. CREATE PIVOT TABLE
# ============================================================

pivot_table = pd.pivot_table(
    df,
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)


# Convert survival rate to percentage
pivot_table = (pivot_table * 100).round(2)


# Rename columns
pivot_table.columns = [
    f"Pclass {col} (%)"
    for col in pivot_table.columns
]


print("2. SURVIVAL RATE BY SEX AND PCLASS")
display(pivot_table)


# ============================================================
# 5. FIND HIGHEST AND LOWEST SURVIVAL GROUP
# ============================================================

# Convert pivot table into a long format
comparison = pivot_table.reset_index().melt(
    id_vars="Sex",
    var_name="Pclass",
    value_name="Survival Rate (%)"
)

# Remove (%) from Pclass names if necessary
comparison["Pclass"] = comparison["Pclass"].str.replace(
    " (%)", "", regex=False
)


# Remove missing values
comparison = comparison.dropna()


# Find highest survival group
highest_group = comparison.loc[
    comparison["Survival Rate (%)"].idxmax()
]


# Find lowest survival group
lowest_group = comparison.loc[
    comparison["Survival Rate (%)"].idxmin()
]


# ============================================================
# 6. DISPLAY COMPARISON TABLE
# ============================================================

print("3. COMPARISON OF SEX-PCLASS GROUPS")

display(comparison)


# ============================================================
# 7. DISPLAY HIGHEST AND LOWEST GROUPS
# ============================================================

result_table = pd.DataFrame({
    "Category": ["Highest Survival", "Lowest Survival"],
    "Sex": [
        highest_group["Sex"],
        lowest_group["Sex"]
    ],
    "Pclass": [
        highest_group["Pclass"],
        lowest_group["Pclass"]
    ],
    "Survival Rate (%)": [
        highest_group["Survival Rate (%)"],
        lowest_group["Survival Rate (%)"]
    ]
})

print("4. HIGHEST AND LOWEST SURVIVAL GROUPS")

display(result_table)
```
